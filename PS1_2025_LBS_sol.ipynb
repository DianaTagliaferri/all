Import relevant libraries

```python
import numpy as np
import scipy.stats
from scipy.optimize import root
```

Numerically solve minimization problem, **Case 1**

```python
def focs(vars: np.array):
    '''
    Args:
    - vars (np.array): A numpy array containing the initial guess for the variables a, b, and x_0

    Returns:
    - An array of length 3 representing the first-order conditions of the minimization problem
    '''

    a, b, x_0 = vars

    # First-order condition with respect to a
    foc_a = 2 * a * scipy.stats.norm.cdf(x_0) + 2 * scipy.stats.norm.pdf(x_0) - (2/(np.sqrt(2 * np.pi)))

    # First-order condition with respect to b
    foc_b = b * scipy.stats.norm.cdf(x_0) + scipy.stats.norm.pdf(x_0) - b

    # First-order condition with respect to x_0
    foc_x0 = (a**2 - b**2) - 2 *x_0*(a-b)

    return np.array([foc_a, foc_b, foc_x0])
```

```python
sol = root(focs, [0.1, 1.3, 0.7])
sol
```

```python
a, b, x_0 = sol.x
a, b, x_0
```

```python
# Evaluate the objective function at the solution
exp = scipy.stats.norm.cdf(x_0) * (a**2 - b**2) + 2*scipy.stats.norm.pdf(x_0) * (a - b) - (2*a/np.sqrt(2 * np.pi)) + b**2
1.5+exp
```
